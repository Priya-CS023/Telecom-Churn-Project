# preprocessing

#replacing 'No internet service' with 'No' from these categorical columns

```python
columns = ['MultipleLines', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']

df[columns] = df[columns].replace('No internet service', 'No')
```

```python
df.head(5)
```

```python
columns = ['MultipleLines']
df[columns] = df[columns].replace('No phone service', 'No')
```

```python
df.head(5)
```

```python
# Preparing the data for modeling
X = df.drop("Churn", axis=1)
y = df["Churn"]
```

```python
y = y.map({"Yes": 1, "No": 0})
```

```python
from sklearn.model_selection import train_test_split
```

```python
# Splitting the data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split( X, y,  test_size=0.20, random_state=42, stratify=y)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
```

```python
# Define the numerical and categorical features
numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges",
    "SeniorCitizen"
]

categorical_features = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingMovies",
    "Contract",
    "PaymentMethod"
]
```

```python
# Define the preprocessing pipelines for numerical and categorical features
numerical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler())
    ]
)

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
# Apply the numerical and categorical preprocessing pipelines to their respective feature columns

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ]
)
```

```python
# Create an end-to-end pipeline that preprocesses features and trains the classifier
model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("classifier", LogisticRegression(max_iter=5000))
    ]
)
```

```python
model.fit(X_train, y_train)
```

```python
# Evaluate the model using 5-fold cross-validation

from sklearn.model_selection import cross_val_score

cv_scores = cross_val_score(
    model,
    X_train,
    y_train,
    cv=5,
    scoring="f1"
)

print("CV scores:", cv_scores)
print("Mean CV F1:", cv_scores.mean())
```

```python
# Define the hyperparameter combinations to tune for Logistic Regression

from sklearn.model_selection import GridSearchCV

param_grid = {
    "classifier__C": [0.01, 0.1, 1, 10, 100],
    "classifier__solver": ["liblinear", "lbfgs"]
}

grid_search = GridSearchCV(
    estimator=model,
    param_grid=param_grid,
    cv=5,
    scoring="f1",
    n_jobs=-1
)

grid_search.fit(X_train, y_train)
```

```python
print("Best parameters:", grid_search.best_params_)
print("Best CV F1:", grid_search.best_score_)
```

```python
# Get the best model from the grid search
best_model = grid_search.best_estimator_
```

```python
#measuring the performance of the best model on the training set
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)

y_train_pred = best_model.predict(X_train)

print("Training Accuracy:", accuracy_score(y_train, y_train_pred))
print("Training Precision:", precision_score(y_train, y_train_pred))
print("Training Recall:", recall_score(y_train, y_train_pred))
print("Training F1:", f1_score(y_train, y_train_pred))
```

```python
# Measure the performance of the best model on the test set
y_test_pred = best_model.predict(X_test)
```

```python
print("Test Accuracy:", accuracy_score(y_test, y_test_pred))
print("Test Precision:", precision_score(y_test, y_test_pred))
print("Test Recall:", recall_score(y_test, y_test_pred))
print("Test F1:", f1_score(y_test, y_test_pred))
```

```python
#measure the performance of the best model on the test set using ROC-AUC score
from sklearn.metrics import roc_auc_score

# Probability of churn (class = 1)
y_test_proba = best_model.predict_proba(X_test)[:, 1]

roc_auc = roc_auc_score(y_test, y_test_proba)

print("Test ROC-AUC:", roc_auc)
```

```python
#auc-roc curve
from sklearn.metrics import roc_curve, roc_auc_score
import matplotlib.pyplot as plt

y_test_proba = best_model.predict_proba(X_test)[:, 1]

roc_auc = roc_auc_score(y_test, y_test_proba)

fpr, tpr, thresholds = roc_curve(y_test, y_test_proba)

plt.figure(figsize=(7, 5))
plt.plot(
    fpr,
    tpr,
    label=f"Logistic Regression (AUC = {roc_auc:.3f})"
)

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
# Generate out-of-fold probability predictions using 5-fold cross-validation.
# Extract the probability of the positive class (Churn = Yes)
from sklearn.model_selection import cross_val_predict

y_train_proba_cv = cross_val_predict(
    best_model,
    X_train,
    y_train,
    cv=5,
    method="predict_proba",
    n_jobs=-1
)[:, 1]
```

```python
# Define thresholds to evaluate between 0.30 and 0.60
import numpy as np
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds_to_test = np.arange(0.30, 0.61, 0.05)

# Store evaluation metrics for each threshold
threshold_results = []

# Evaluate model performance at each classification threshold
for threshold in thresholds_to_test:

    # Convert predicted probabilities into binary predictions
    # using the current threshold
    y_pred_threshold = (
        y_train_proba_cv >= threshold
    ).astype(int)

    # Calculate precision, recall, and F1 score
    precision = precision_score(
        y_train,
        y_pred_threshold
    )

    recall = recall_score(
        y_train,
        y_pred_threshold
    )

    f1 = f1_score(
        y_train,
        y_pred_threshold
    )

    # Store the results for the current threshold
    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1": f1
    })

# Convert the threshold results into a DataFrame for comparison
threshold_df = pd.DataFrame(threshold_results)

print(threshold_df)
```

```python
final_threshold = 0.30
```

```python
# Apply the final threshold to the test set probabilities
y_test_proba = best_model.predict_proba(X_test)[:, 1]

y_test_final = (
    y_test_proba >= final_threshold
).astype(int)
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("Final Threshold:", final_threshold)

print("Test Accuracy:",
      accuracy_score(y_test, y_test_final))

print("Test Precision:",
      precision_score(y_test, y_test_final))

print("Test Recall:",
      recall_score(y_test, y_test_final))

print("Test F1:",
      f1_score(y_test, y_test_final))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_test_final))
```

```python
# Visualize the confusion matrix using a heatmap
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_test_final)

plt.figure(figsize=(6, 5))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    xticklabels=["No Churn", "Churn"],
    yticklabels=["No Churn", "Churn"]
)

plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix — Threshold 0.30")

plt.show()
```

```python
# Analyze the coefficients of the logistic regression model
#To determine the most important features contributing to churn prediction
logistic_model = best_model.named_steps["classifier"]

feature_names = best_model.named_steps[
    "preprocessor"
].get_feature_names_out()

coefficients = logistic_model.coef_[0]

coefficient_df = pd.DataFrame({
    "Feature": feature_names,
    "Coefficient": coefficients
})

coefficient_df["Absolute_Coefficient"] = (
    coefficient_df["Coefficient"].abs()
)

coefficient_df = coefficient_df.sort_values(
    "Absolute_Coefficient",
    ascending=False
)

print(coefficient_df.head(15))
```
